```python
import numpy as np
```

```python
a = np.array([1, 2, 3])
print(a)
```

```python
b = np.array([[1, 2, 3],
              [4, 5, 6]])
print(b)
```

```python
print(b.shape)   # (2, 3)
print(b.dtype)   # int64
```

```python
a = np.array([1, 2, 3])

print(a + 1)      # [2 3 4]
print(a * 2)      # [2 4 6]
```

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print(x + y)      # [5 7 9]
```

```python
print(np.zeros((2, 3)))   # 0으로 채움
print(np.ones((2, 3)))    # 1로 채움
print(np.arange(0, 10))   # 0~9
print(np.linspace(0, 1, 5))  # 0~1 사이 5개
```

```python
a = np.array([10, 20, 30])

print(a[0])     # 10
print(a[1:3])   # [20 30]
```

```python
b = np.array([[1,2,3],[4,5,6]])

print(b[1,1])   # 2
```
