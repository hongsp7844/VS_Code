# 📌 Pandas 전처리 실습 1
## 🎯 문제
결측치 처리  
- name None → 삭제  
- age 평균값 채우기  
- salary 평균값 채우기  

데이터 타입 확인  
- 각 컬럼 dtype 출력  

컬럼 생성  
- salary ≥ 55000 → "high"  
- salary < 55000 → "low"  
→ salary_level 생성  

필터링  
- age ≥ 25 데이터만 출력  


```python
import pandas as pd

data = {
    "name": ["Alice", "Bob", "Charlie", "David", None],
    "age": [25, None, 30, 22, 28],
    "salary": [50000, 60000, None, 52000, 58000]
}

df = pd.DataFrame(data)
df
```

```python
df.dropna(subset=['name'], inplace=True)
df['age']=df['age'].fillna(df['age'].mean())
df['salary']=df['salary'].fillna(df['salary'].mean())
print(df.dtypes)
df['salary_level']=df["salary"].apply(lambda x: "high" if x >= 55000 else "low")
cond=df['age']>=25
result = df[cond]
print(result)
```

# 📌 Pandas 전처리 실습 2

## 🎯 문제

결측치 처리  
- name None → 삭제  
- age 평균값 채우기  
- salary 평균값 채우기  

중복 제거  
- 전체 행 기준 중복 제거  

컬럼 생성  
- age 기준  
  - 25 이상 → "adult"  
  - 25 미만 → "young"  
→ age_group 생성  

필터링  
- salary ≥ 35000 인 데이터만 출력  

```python
import pandas as pd
data = {
    "name": ["Tom", "Jane", "Mike", None, "Anna", "Tom"],
    "age": [20, 25, None, 30, 22, 20],
    "salary": [30000, None, 40000, 50000, None, 30000]
}
df = pd.DataFrame(data)
```

```python
df.dropna(subset=['name'],inplace=True)
df['age']=df['age'].fillna(df['age'].mean())
df['salary']=df['salary'].fillna(df['salary'].mean())
df.drop_duplicates(inplace=True)
df['age_group']=df['age'].apply(lambda x: "adult" if x >= 25 else "young")
result=df[df['salary']>=35000]
print(result)
```

# 📌 Pandas 전처리 실습 3

## 🎯 문제

결측치 처리  
- name None → 삭제  
- age 평균값 채우기  
- salary 평균값 채우기  

중복 제거  
- name 기준 중복 제거 (첫 번째만 유지)  

컬럼 생성  
- salary 기준  
  - 40000 이상 → "high"  
  - 40000 미만 → "low"  
→ salary_level 생성  

필터링  
- age ≥ 23 이고 salary ≥ 40000 인 데이터만 출력  

```python
import pandas as pd

data = {
    "name": ["Alex", "Bob", "Alex", "Chris", None, "Bob"],
    "age": [22, None, 22, 27, 25, None],
    "salary": [38000, 42000, 38000, None, 41000, 42000]
}

df = pd.DataFrame(data)
```

```python
df.dropna(subset=['name'],inplace=True)
df['age']=df['age'].fillna(df['age'].mean())
df['salary']=df['salary'].fillna(df['salary'].mean())
df.drop_duplicates(subset=['name'],inplace=True)
df['salary_level']=df['salary'].apply(lambda x: 'high' if x >=40000 else 'low')
result=df[(df['age']>=23) & (df['salary']>=40000)]
print(result)
```

# 📌 Pandas 전처리 실습 2-3

## 🎯 문제

결측치 처리  
- product None → 삭제  
- quantity 평균값 채우기  
- price 평균값 채우기  

중복 제거  
- product 기준 중복 제거 (첫 번째만 유지)  

컬럼 생성  
- price 기준  
  - 5000 이상 → "expensive"  
  - 5000 미만 → "cheap"  
→ price_level 생성  

필터링  
- quantity ≥ 5 이고 price ≥ 5000 인 데이터만 출력  

```python
import pandas as pd

data = {
    "product": ["Apple", "Banana", "Apple", "Orange", None, "Banana"],
    "quantity": [10, None, 10, 3, 7, None],
    "price": [3000, 6000, 3000, None, 5500, 6000]
}

df = pd.DataFrame(data)
df
```

```python
df.dropna(subset=['product'],inplace=True)
df['quantity']=df['quantity'].fillna(df['quantity'].mean())
df['price']=df['price'].fillna(df['price'].mean())
df.drop_duplicates(subset=['product'], inplace=True)
df['price_level']=df['price'].apply(lambda x: 'expensive' if x>=5000 else 'cheap')
result=df[(df['quantity'] >= 5) & (df['price'] >= 5000)]
print(result)
```
