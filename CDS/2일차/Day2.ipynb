```python
import pandas as pd

df = pd.DataFrame({
    '이름':['김철수','이영희','박민수','최지은','한수지'],
    '나이':[25,31,28,22,35],
    '급여':[3000,4500,3800,2800,5200],
    '부서':['개발','인사','개발','영업','개발']
})

df
```

```python
df[df['급여']>=3000]
df[df['나이']>=30]
df[df['부서']=='개발']
df[(df['부서']=='개발')&(df['급여']>=4000)]
df.sort_values('급여',ascending=False)
df.sort_values('나이',ascending=True)
df.drop(columns=['부서'])
df.drop(index=0)
df.loc[df['이름']=='최지은','급여']=3200
df.loc[df['부서']=='영업','보너스']=700
```

```python
df = df[df['급여']>=3000]
df=df[df['부서']=='개발']
df=df.sort_values('급여',ascending=False)
df['보너스']=1000
df = df.drop(columns='나이')
df
```
